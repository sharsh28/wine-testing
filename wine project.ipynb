# ***WINE QUALITY PREDITION***

**IMPORT REQUIRED LIBRARY**

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

from warnings import filterwarnings
filterwarnings(action='ignore')
```

# **LOADING DATASET**

```python
wine=pd.read_csv("/content/WineQT.csv")
print("Successfully import data")
print(wine)
```

```python
sns.heatmap(wine.isnull())
```

# **DISPLOT**

```python
sns.displot(wine['alcohol'])
```

# **HISTOGRAM**

```python
wine.hist(figsize=(10,10),bins=50)
plt.show()
```

# **PAIR PLOT**

```python
sns.pairplot(wine,hue="quality")
```

```python
x=wine.iloc[:,:-1].values
y=wine.iloc[:,-1].values
```

```python
print(y)
```

# **FEATURES IMPORTANCES**

**SPLITTING DATASET**

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test=train_test_split(x,y,test_size=0.3)
```

**USING LOGISTIC REGRESSION**




```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
from sklearn.metrics import accuracy_score,confusion_matrix
print("Accuracy score:",accuracy_score(y_test,y_pred))
```

```python
from sklearn.metrics import classification_report
cr=classification_report(y_test,y_pred)
print(cr)
```

**USING KNN ALGORITHM**

```python
from sklearn.neighbors import KNeighborsClassifier
model=KNeighborsClassifier(n_neighbors=3)
model.fit(x_train,y_train)
y_pred=model.predict(x_test)
from sklearn.metrics import accuracy_score
print("Accuracy Score:",accuracy_score(y_test,y_pred))
```

**USING GAUSSIAN NB**



```python
from sklearn.naive_bayes import GaussianNB
model3 = GaussianNB()
model3.fit(x_train,y_train)
y_pred3=model3.predict(x_test)
from sklearn.metrics import accuracy_score
print("Accuracy score:",accuracy_score(y_test,y_pred3))
```

**DECISION TREE ALGORITHM**


```python
from sklearn.tree import DecisionTreeClassifier
dt=DecisionTreeClassifier()
dt.fit(x_train,y_train)
```

```python
ypred=dt.predict(x_test)
ac=accuracy_score(y_test,ypred)
print(ac)
```

**RANDOM FOREST ALGORITHM**


```python
from sklearn.ensemble import RandomForestClassifier
rf=RandomForestClassifier()
rf.fit(x_train,y_train)
```

```python
ypred=dt.predict(x_test)
ac=accuracy_score(y_test,ypred)
print(ac)
```

**SVM LINEAR ALGORITHM**


```python
from sklearn.svm import SVC
svm_model = SVC(kernel='linear')
svm_model.fit(x_train, y_train)
```

```python
ypred=svm_model.predict(x_test)
ac=accuracy_score(y_test,ypred)
print(ac)
```
